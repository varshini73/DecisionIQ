# 01 — Exploratory Data Analysis

Purpose: understand the raw retail dataset before any modeling.
Checks: date range, store/SKU counts, missing values, sales distribution,
stockout frequency, seasonality patterns, price/promo coverage.

Run `src/data_preprocessing.py` first, or point this notebook at
`data/raw/retail_inventory_raw.csv` directly for pre-cleaning exploration.

```python
import sys
sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from utils import DATA_RAW_DIR, DATA_PROCESSED_DIR

df = pd.read_parquet(DATA_PROCESSED_DIR / 'clean_panel.parquet')
df.head()
```

```python
print('Date range:', df['date'].min(), '->', df['date'].max())
print('Stores:', df['store_id'].nunique(), '| SKUs:', df['sku_id'].nunique())
print('Stockout day rate:', df['is_stockout_day'].mean())
df.isna().mean().sort_values(ascending=False).head(10)
```

```python
daily_sales = df.groupby('date')['units_sold'].sum()
daily_sales.plot(figsize=(12,4), title='Total Units Sold Over Time')
plt.show()
```

```python
sns.histplot(df['units_sold'], bins=50)
plt.title('Distribution of Daily Units Sold')
plt.show()
```
