# 02 — Feature Engineering Walkthrough

Purpose: visually validate the features created in `src/feature_engineering.py`
before they're fed to the forecasting models -- lags, rolling stats,
demand un-censoring, ABC/XYZ classification.

```python
import sys
sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
from utils import DATA_PROCESSED_DIR

df = pd.read_parquet(DATA_PROCESSED_DIR / 'features.parquet')
df[['date','store_id','sku_id','units_sold','demand_est','is_stockout_day']].head(10)
```

```python
# Visualize the demand un-censoring correction on a stockout-heavy SKU
sample = df[df['is_stockout_day']==1][['store_id','sku_id']].drop_duplicates().iloc[0]
s = df[(df.store_id==sample.store_id)&(df.sku_id==sample.sku_id)].sort_values('date')
plt.figure(figsize=(12,4))
plt.plot(s['date'], s['units_sold'], label='Observed units_sold (censored)')
plt.plot(s['date'], s['demand_est'], label='Corrected demand_est', linestyle='--')
plt.legend(); plt.title('Demand Un-censoring Example')
plt.show()
```
