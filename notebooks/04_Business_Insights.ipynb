# 04 — Business Insights

Purpose: translate model + optimization outputs into the business-facing
story -- total $ at risk, stockout/overstock rates by ABC class, and
top SKUs to act on this week. This is the notebook you'd screenshot
for a stakeholder deck.

```python
import sys
sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
from utils import DATA_PROCESSED_DIR

decisions = pd.read_parquet(DATA_PROCESSED_DIR / 'decisions.parquet')
decisions['decision'].value_counts()
```

```python
total_at_risk = decisions['estimated_stockout_cost'].sum()
total_holding = decisions['annual_holding_cost'].sum()
print('Total $ at risk from stockouts:', round(total_at_risk, 2))
print('Total annual holding cost tied up:', round(total_holding, 2))
```

```python
decisions.groupby('abc_class')['decision'].value_counts(normalize=True).unstack().plot(
    kind='bar', stacked=True, figsize=(8,5), title='Decision Mix by ABC Class')
plt.show()
```

```python
# Top 10 SKUs to act on this week, ranked by $ impact
decisions.sort_values('estimated_stockout_cost', ascending=False).head(10)[
    ['store_id','sku_id','decision','priority','recommended_qty','estimated_stockout_cost']]
```
