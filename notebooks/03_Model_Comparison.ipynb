# 03 — Model Comparison

Purpose: compare Seasonal Naive vs. LightGBM/XGBoost vs. Ensemble on
WAPE, RMSE, Bias, and Forecast Value Added (FVA). Run
`src/demand_forecasting.py` first to generate `model_comparison.csv`.

```python
import sys
sys.path.append('../src')
import pandas as pd
import matplotlib.pyplot as plt
from utils import DATA_PROCESSED_DIR

results = pd.read_csv(DATA_PROCESSED_DIR / 'model_comparison.csv')
results
```

```python
results.plot(x='model', y='WAPE', kind='bar', title='WAPE by Model (lower is better)')
plt.show()
```

```python
# Forecast Value Added -- the business-readable 'is this model worth it' metric
if 'FVA_vs_baseline' in results.columns:
    results.dropna(subset=['FVA_vs_baseline']).plot(
        x='model', y='FVA_vs_baseline', kind='bar',
        title='Forecast Value Added vs. Seasonal Naive Baseline')
    plt.show()
```
